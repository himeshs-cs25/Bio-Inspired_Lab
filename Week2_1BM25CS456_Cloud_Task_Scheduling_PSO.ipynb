```python

import random

# Cloud tasks: execution length in Million Instructions (MI)
tasks = [10, 20, 30, 40, 50]

# Virtual machines: processing capacity in MIPS
vms = [10, 20, 30]

# PSO parameters
particles = 10
max_iterations = 50
w = 0.5
c1 = 1.5
c2 = 1.5

num_tasks = len(tasks)
num_vms = len(vms)


# Calculate makespan (fitness function)
def fitness(position):
    load = [0] * num_vms

    for i in range(num_tasks):
        vm = position[i]
        load[vm] += tasks[i] / vms[vm]

    return max(load)


# Initialize particles
positions = [
    [random.randint(0, num_vms - 1)
     for _ in range(num_tasks)]
    for _ in range(particles)
]

velocities = [
    [random.uniform(-1, 1) for _ in range(num_tasks)]
    for _ in range(particles)
]

pbest = [p[:] for p in positions]
pbest_fitness = [fitness(p) for p in positions]

best_index = pbest_fitness.index(min(pbest_fitness))
gbest = pbest[best_index][:]
gbest_fitness = pbest_fitness[best_index]


# PSO optimization
for iteration in range(max_iterations):

    for p in range(particles):
        for i in range(num_tasks):

            r1 = random.random()
            r2 = random.random()

            velocities[p][i] = (
                w * velocities[p][i]
                + c1 * r1 * (pbest[p][i] - positions[p][i])
                + c2 * r2 * (gbest[i] - positions[p][i])
            )

            positions[p][i] += round(velocities[p][i])

            # Keep VM assignment valid
            positions[p][i] = max(
                0, min(num_vms - 1, positions[p][i])
            )

        current_fitness = fitness(positions[p])

        if current_fitness < pbest_fitness[p]:
            pbest[p] = positions[p][:]
            pbest_fitness[p] = current_fitness

        if current_fitness < gbest_fitness:
            gbest = positions[p][:]
            gbest_fitness = current_fitness


# Display best schedule
print("Best Task Scheduling using PSO")
print("--------------------------------")

for i in range(num_tasks):
    print(f"Task T{i + 1} -> VM{gbest[i] + 1}")

print(f"\nMinimum Makespan: {gbest_fitness:.2f} time units")

print("\nVM Workloads:")
for j in range(num_vms):
    workload = sum(
        tasks[i] / vms[j]
        for i in range(num_tasks)
        if gbest[i] == j
    )
    print(f"VM{j + 1}: {workload:.2f} time units")
```
