```python
import random

n = int(input("Enter number of agricultural regions: "))

distance = []

print("Enter distance matrix:")

for i in range(n):
    row = list(map(int, input().split()))
    distance.append(row)

ants = 10
iterations = 50
alpha = 1
beta = 2
evaporation = 0.5

pheromone = [[1 for j in range(n)] for i in range(n)]

best_path = None
best_distance = float('inf')

for _ in range(iterations):

    for ant in range(ants):

        path = [0]
        unvisited = list(range(1, n))

        while unvisited:
            current = path[-1]
            probabilities = []

            for city in unvisited:
                p = pheromone[current][city] ** alpha * \
                    (1 / distance[current][city]) ** beta
                probabilities.append(p)

            total = sum(probabilities)
            probabilities = [p / total for p in probabilities]

            next_city = random.choices(
                unvisited, weights=probabilities
            )[0]

            path.append(next_city)
            unvisited.remove(next_city)

        path.append(0)

        total_distance = sum(
            distance[path[i]][path[i + 1]]
            for i in range(n)
        )

        if total_distance < best_distance:
            best_distance = total_distance
            best_path = path[:]

    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - evaporation)

    for i in range(n):
        for j in range(i + 1, n):
            pheromone[i][j] += 1 / best_distance
            pheromone[j][i] = pheromone[i][j]

print("\nOptimized Agricultural Drone Path:")
print(" -> ".join(map(str, best_path)))
print("Minimum Distance:", best_distance)
```
