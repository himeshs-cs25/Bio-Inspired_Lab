```python
import random

n = int(input("Enter number of possible locations: "))
cost = list(map(int, input("Enter installation costs: ").split()))
demand = list(map(int, input("Enter EV demand at each location: ").split()))

nests = 10
iterations = 50
Pa = 0.25

population = [
    [random.randint(0, 1) for _ in range(n)]
    for _ in range(nests)
]

def fitness(solution):
    total_cost = sum(solution[i] * cost[i] for i in range(n))
    coverage = sum(solution[i] * demand[i] for i in range(n))
    if total_cost == 0:
        return 0
    return coverage / total_cost

best = max(population, key=fitness)

for _ in range(iterations):
    for i in range(nests):
        new = population[i][:]
        j = random.randrange(n)
        new[j] = 1 - new[j]

        if fitness(new) > fitness(population[i]):
            population[i] = new

        if fitness(population[i]) > fitness(best):
            best = population[i][:]

    for i in range(nests):
        if random.random() < Pa:
            population[i] = [
                random.randint(0, 1) for _ in range(n)
            ]

    candidate = max(population, key=fitness)
    if fitness(candidate) > fitness(best):
        best = candidate[:]

print("\nOptimal Charging Station Locations:")
for i in range(n):
    if best[i] == 1:
        print("Location", i + 1)

print("Maximum Fitness:", round(fitness(best), 4))
```
