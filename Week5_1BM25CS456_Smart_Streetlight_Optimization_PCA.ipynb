```python

import multiprocessing
import random

# Function to optimize each streetlight cell
def optimize_light(data):
    traffic, old_intensity, neighbors = data

    if traffic >= 70:
        intensity = 100
    elif traffic >= 40:
        intensity = 70
    elif traffic >= 20:
        intensity = 50
    else:
        intensity = 30

    # Consider neighbouring streetlight conditions
    if neighbors >= 70 and intensity < 50:
        intensity = 50

    energy = intensity / 100

    return intensity, energy


if __name__ == "__main__":

    n = int(input("Enter number of streetlights: "))

    traffic = []
    intensity = []

    print("\nEnter traffic density (0-100):")

    for i in range(n):
        t = int(input(f"Streetlight {i+1}: "))
        traffic.append(max(0, min(100, t)))
        intensity.append(100)

    # Create cellular data with neighbouring traffic
    data = []

    for i in range(n):
        left = traffic[i-1] if i > 0 else 0
        right = traffic[i+1] if i < n-1 else 0

        neighbors = max(left, right)

        data.append((traffic[i], intensity[i], neighbors))

    # Parallel processing
    with multiprocessing.Pool() as pool:
        results = pool.map(optimize_light, data)

    print("\n--- OPTIMIZED STREETLIGHTS ---")

    total_energy = 0

    for i, (new_intensity, energy) in enumerate(results):
        total_energy += energy
        print(f"Streetlight {i+1}: {new_intensity}% intensity")

    print("\nTotal Energy Consumption:", round(total_energy, 2))
    print("Average Energy Consumption:",
          round(total_energy / n, 2))

    print("\nOptimization Completed Successfully!")
```
